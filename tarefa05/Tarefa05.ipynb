```python
import matplotlib.pyplot as plt
import seaborn as sns

# Create a boxplot to visualize the variability of 'Proteína (g)' by 'Categoria'
plt.figure(figsize=(12, 7))
sns.boxplot(x='Categoria', y='Proteína (g)', data=df, palette='viridis')
plt.title('Variabilidade do Teor de Proteína (g) por Categoria')
plt.xlabel('Categoria de Alimento')
plt.ylabel('Proteína (g)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df['Gordura total (g)'], bins=30, kde=True, color='skyblue')
plt.title('Distribuição da Gordura total (g)')
plt.xlabel('Gordura total (g)')
plt.ylabel('Frequência')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate mean and median for 'Gordura total (g)'
mean_fat = df['Gordura total (g)'].mean()
median_fat = df['Gordura total (g)'].median()

fig, axes = plt.subplots(2, 1, figsize=(10, 10), gridspec_kw={'height_ratios': [0.7, 0.3]})

# Histogram for 'Gordura total (g)'
sns.histplot(df['Gordura total (g)'], bins=30, kde=True, ax=axes[0], color='skyblue')
axes[0].axvline(mean_fat, color='red', linestyle='dashed', linewidth=2, label=f'Média: {mean_fat:.2f}')
axes[0].axvline(median_fat, color='green', linestyle='dashed', linewidth=2, label=f'Mediana: {median_fat:.2f}')
axes[0].set_title('Histograma de Gordura total (g)')
axes[0].set_xlabel('Gordura total (g)')
axes[0].set_ylabel('Frequência')
axes[0].legend()
axes[0].grid(axis='y', linestyle='--', alpha=0.7)

# Boxplot for 'Gordura total (g)'
sns.boxplot(x=df['Gordura total (g)'], ax=axes[1], color='lightgreen')
axes[0].axvline(mean_fat, color='red', linestyle='dashed', linewidth=2, label=f'Média: {mean_fat:.2f}')

axes[1].set_title('Boxplot de Gordura total (g)')
axes[1].set_xlabel('Gordura total (g)')
axes[0].axvline(mean_fat, color='red', linestyle='dashed', linewidth=2, label=f'Média: {mean_fat:.2f}')

axes[1].grid(axis='x', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

```python
# Calculate the overall standard deviation of 'Proteína (g)'
overall_protein_std = df['Proteína (g)'].std()
print(f"O desvio padrão total do teor de proteína é: {overall_protein_std:.2f}\n")

# Calculate the standard deviation of 'Proteína (g)' for each 'Categoria'
protein_std_by_category = df.groupby('Categoria')['Proteína (g)'].std().sort_values(ascending=False)

# Display the standard deviation for each category
print("Desvio padrão do teor de proteína por Categoria:")
display(protein_std_by_category)
```
