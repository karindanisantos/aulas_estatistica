### Explicação da Visualização: Gráfico de Pizza da Variável 'categoria'

A variável **'categoria'** foi escolhida para a visualização em gráfico de pizza por ser uma **variável qualitativa nominal**. Isso significa que ela representa diferentes grupos ou tipos de alimentos (e.g., "Carnes e derivados", "Cereais e derivados") que não possuem uma ordem ou hierarquia intrínseca. Para esse tipo de variável, o gráfico de pizza é uma ferramenta visual ideal, pois permite exibir a **distribuição proporcional** das diferentes categorias, mostrando a participação de cada uma no total.

1.  **Visão Geral da Composição do Banco de Dados **: O gráfico de pizza oferece uma visão imediata da **estrutura do BD em termos de tipos de alimentos**. Podemos rapidamente identificar quais categorias são mais ou menos representadas na base de dados TACO.

Por exemplo, se uma categoria como 'Gorduras e óleos' (que veremos que se mostrou rica em lipídeos) representa uma pequena fatia do gráfico, sabemos que sua influência nos dados gerais pode ser restrita ao seu número de amostras, embora impactante na média. Inversamente, se uma categoria majoritária tem valores médios de lipídeos ou energia, isso tem um peso maior na composição nutricional geral dos alimentos da TACO.

3.  **Identificação de Foco**: O gráfico ajuda a perceber quais categorias de alimentos dominam o banco de dados, o que pode direcionar futuras análises mais aprofundadas ou indicar onde concentrar esforços na avaliação nutricional.

Em resumo, o gráfico de pizza da 'categoria' oferece uma perspectiva macro da diversidade e representatividade dos alimentos no dataset, servindo como um ponto de partida fundamental para entender a composição global antes de mergulhar em análises nutricionais mais específicas e quantitativas.

### Diferença entre Gráfico de Pizza e Gráfico de Frequência para a Variável 'categoria'

Ambos os gráficos utilizam a variável 'categoria' para visualização, mas com propósitos e focos distintos:

**1. Gráfico de Pizza (Gráfico de Setores):**

*   **Propósito:** O gráfico de pizza (como o que geramos para a 'categoria') é ideal para mostrar a **proporção de cada parte em relação a um todo**. Ele divide um círculo em 'fatias', onde o tamanho de cada fatia é proporcional à porcentagem que aquela categoria representa do total.
*   **O que ele mostra:** Ele destaca a **composição proporcional** do conjunto de dados. Por exemplo, você pode ver rapidamente qual categoria de alimento representa a maior ou menor porcentagem do total de itens na base de dados TACO.
*   **Melhor Uso:** Comparar a participação relativa de poucas categorias (idealmente até 5-7 categorias para clareza) dentro de um total, enfatizando a 'fatia' de cada uma.

**2. Gráfico de Frequência (Gráfico de Barras de Frequência):**

*   **Propósito:** Um gráfico de frequência (geralmente um gráfico de barras) é usado para exibir a **contagem absoluta ou a frequência** com que cada categoria aparece nos dados.
*   **O que ele mostra:** Ele permite visualizar diretamente a **quantidade** de ocorrências de cada categoria. Você veria uma barra para cada categoria, e a altura da barra indicaria quantos alimentos pertencem àquela categoria.
*   **Melhor Uso:** Comparar o número exato de itens em cada categoria, ou seja, as frequências. É mais eficaz para conjuntos de dados com um número maior de categorias, onde as proporções em um gráfico de pizza poderiam se tornar difíceis de ler.

**Em Resumo:**

*   O **Gráfico de Pizza** enfatiza a **proporção** (`% do total`).
*   O **Gráfico de Frequência (Barras)** enfatiza a **contagem** ou **quantidade** de cada categoria.

Ambos são válidos e complementares. O gráfico de pizza te dá uma visão rápida da 'distribuição percentual', enquanto o gráfico de barras te dá a 'distribuição de contagens' e é geralmente mais fácil de ler para um número maior de categorias ou para comparações exatas de magnitude entre elas.

### Calculando Quartis e Amplitude Interquartil para Lipídeos

Antes de visualizar a distribuição dos lipídeos através do boxplot, é útil calcular algumas medidas de posição e dispersão, como os quartis (Q1, Q2 e Q3) e a Amplitude Interquartil (AIQ). Essas métricas fornecem uma base numérica para entender a centralidade, a dispersão e a possível presença de outliers na distribuição dos dados.

```python
Q1_lipideos_pre = df['lipideos_g'].quantile(0.25)
Q2_lipideos_pre = df['lipideos_g'].quantile(0.50) # Mediana
Q3_lipideos_pre = df['lipideos_g'].quantile(0.75)

print(f"Primeiro Quartil (Q1) de lipídeos: {Q1_lipideos_pre:.2f} g/100g")
print(f"Mediana (Q2) de lipídeos: {Q2_lipideos_pre:.2f} g/100g")
print(f"Terceiro Quartil (Q3) de lipídeos: {Q3_lipideos_pre:.2f} g/100g")
```

```python
IQR_lipideos_pre = Q3_lipideos_pre - Q1_lipideos_pre
print(f"Amplitude Interquartílica (IQR) de lipídeos: {IQR_lipideos_pre:.2f} g/100g")
```

Os outliers (valores atípicos) presentes no boxplot de lipideos_g representam alimentos que possuem uma quantidade de lipídios significativamente maior do que a maioria dos outros alimentos no dataset.

No boxplot, os pontos individuais que aparecem acima (ou abaixo, se houvesse) dos 'bigodes' são os outliers. Eles são identificados como valores que estão a mais de 1.5 vezes o Intervalo Interquartil (IQR) acima do terceiro quartil (Q3) ou abaixo do primeiro quartil (Q1).

O que isso significa para lipideos_g?

Alimentos com Lipídios Excepcionalmente Altos: A presença desses pontos indica que existem alguns alimentos na base de dados com um teor de lipídios muito elevado em comparação com a grande maioria dos outros alimentos.
Potenciais Alimentos Ricos em Gordura: Esses podem ser produtos como óleos, algumas carnes processadas, nozes ou sementes, que naturalmente contêm uma concentração muito alta de gorduras.

```python
Q1 = df['lipideos_g'].quantile(0.25)
Q3 = df['lipideos_g'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df['lipideos_g'] < lower_bound) | (df['lipideos_g'] > upper_bound)]

print(f"Limites para Outliers (lipideos_g):\nInferior: {lower_bound:.2f}\nSuperior: {upper_bound:.2f}")
print("\nAlimentos identificados como outliers (descricao e lipideos_g):")
display(outliers[['descricao', 'lipideos_g', 'categoria']].sort_values(by='lipideos_g', ascending=False))
```

Para identificar os alimentos que são considerados outliers no boxplot de lipídeos, foi calculado os limites inferior e superior usando o método do Intervalo Interquartil (IQR).

A metodologia para identificação de outliers é a seguinte:

### Fórmulas para Cálculo de Outliers

1.  **Amplitude Interquartil (AIQ)**:
    $$ AIQ = Q3 - Q1 $$

2.  **Limite Inferior (LI)**:
    $$ LI = Q1 - 1.5 \times AIQ $$

3.  **Limite Superior (LS)**:
    $$ LS = Q3 + 1.5 \times AIQ $$

Onde:
*   **Q1** é o Primeiro Quartil (25º percentil).
*   **Q3** é o Terceiro Quartil (75º percentil).

Valores que estão abaixo do Limite Inferior ou acima do Limite Superior são considerados outliers.

### Quartis da Distribuição de Lipídeos

Os quartis são medidas de posição que dividem um conjunto de dados ordenado em quatro partes iguais. Eles são fundamentais para entender a distribuição dos dados, especialmente no contexto de um boxplot:

Primeiro Quartil (Q1): Representa o valor abaixo do qual 25% dos dados estão localizados. No nosso caso, Q1 para lipideos_g é 0.21 g/100g. Isso significa que 25% dos alimentos na TACO têm até 0.21 gramas de lipídeos por 100 gramas.
Segundo Quartil (Q2) ou Mediana: Representa o valor abaixo do qual 50% dos dados estão localizados. É o ponto central da distribuição. Para lipideos_g, a mediana (Q2) é 2.33 g/100g. Isso nos diz que metade dos alimentos tem menos de 2.33 g de lipídeos/100g e a outra metade tem mais.
Terceiro Quartil (Q3): Representa o valor abaixo do qual 75% dos dados estão localizados. Para lipideos_g, Q3 é 11.96 g/100g. Isso indica que 75% dos alimentos têm até 11.96 gramas de lipídeos por 100 gramas.

```python
Q1_lipideos = df['lipideos_g'].quantile(0.25)
Q2_lipideos = df['lipideos_g'].quantile(0.50) # Mediana
Q3_lipideos = df['lipideos_g'].quantile(0.75)

print(f"Primeiro Quartil (Q1) de lipídeos: {Q1_lipideos:.2f} g/100g")
print(f"Mediana (Q2) de lipídeos: {Q2_lipideos:.2f} g/100g")
print(f"Terceiro Quartil (Q3) de lipídeos: {Q3_lipideos:.2f} g/100g")
```

O que os quartis nos revelam:

Dispersão dos dados: A distância entre Q1 e Q3 (a Amplitude Interquartílica - IQR) nos dá uma ideia da dispersão dos 50% centrais dos dados. Um IQR grande indica maior variabilidade, enquanto um IQR pequeno sugere que os dados estão mais agrupados em torno da mediana.
Assimetria: A posição da mediana dentro da 'caixa' do boxplot (entre Q1 e Q3) pode indicar se a distribuição é simétrica ou assimétrica. Se a mediana estiver mais próxima de Q1, a distribuição é assimétrica à direita; se estiver mais próxima de Q3, é assimétrica à esquerda.
Outliers: Os quartis são usados para calcular os limites para identificação de outliers. Valores que estão muito além de Q1 - 1.5 * IQR ou Q3 + 1.5 * IQR são considerados outliers, pois se desviam significativamente da massa central dos dados

Gráfico de dispersão para verificar a relação entre o teor de lipídeos e o valor energético dos alimentos e visualizar a correlação entre duas variáveis quantitativas.

```python
import pandas as pd
from scipy.stats import linregress
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("taco_composicao.csv")

# Realizar a regressão linear
# Removendo NaNs de ambas as colunas para o cálculo da regressão
lipideos_validos = df['lipideos_g'].dropna()
energia_validos = df['energia_kcal'].dropna()

# Certificar-se de que ambas as séries têm o mesmo índice para alinhamento correto
dados_regressao = pd.DataFrame({'lipideos_g': lipideos_validos, 'energia_kcal': energia_validos}).dropna()

slope, intercept, r_value, p_value, std_err = linregress(dados_regressao['lipideos_g'], dados_regressao['energia_kcal'])

print(f"Coeficiente Angular (Slope): {slope:.2f}")
print(f"Intercepto: {intercept:.2f}")
print(f"Valor de R da Regressão Linear: {r_value:.2f}")
# Definir correlacao_pearson para ser usada em outras células (se necessário)
correlacao_pearson = dados_regressao['lipideos_g'].corr(dados_regressao['energia_kcal'])
print(f"Coeficiente de Correlação de Pearson: {correlacao_pearson:.2f}")
```

```python
# Adicionar a reta de regressão ao gráfico de dispersão
plt.figure(figsize=(10, 6))
sns.scatterplot(x='lipideos_g', y='energia_kcal', data=df, alpha=0.6)
plt.plot(dados_regressao['lipideos_g'], intercept + slope * dados_regressao['lipideos_g'], color='red', label='Reta de Regressão Estimada')
plt.title('Relação entre Lipídeos e Energia em Alimentos (TACO) com Reta de Regressão')
plt.xlabel('Lipídeos (g/100g)')
plt.ylabel('Energia (kcal/100g)')
plt.grid(True, linestyle='--', alpha=0.7)
plt.legend()
plt.tight_layout()
plt.show()
```

*   **Coeficiente Angular (Slope - $b_1$):** 8.39
    *   **Interpretação:** O coeficiente angular de **8.39** significa que, para cada aumento de 1 grama de lipídeos por 100 gramas de alimento (`lipideos_g`), espera-se um aumento médio de **8.39** kcal no valor energético (`energia_kcal`), mantendo outras variáveis constantes. ## Correlação Linear de Pearson

O coeficiente de correlação de Pearson (r) mede a força e a direção de uma relação linear entre duas variáveis quantitativas. Seu valor varia de -1 a +1:

*   **r = +1**: Indica uma perfeita correlação linear positiva (à medida que uma variável aumenta, a outra também aumenta de forma consistente).
*   **r = -1**: Indica uma perfeita correlação linear negativa (à medida que uma variável aumenta, a outra diminui de forma consistente).
*   **r = 0**: Indica que não há relação linear entre as variáveis.

### Interpretação da Correlação Encontrada

Como calculado anteriormente, o coeficiente de correlação de Pearson entre `lipideos_g` e `energia_kcal` é de **0.81**. Isso reitera que alimentos com maior quantidade de lipídeos tendem a apresentar maior valor energético.

### Correlação Próxima a Zero

Uma correlação linear de Pearson igual ou próxima de zero indica a ausência de uma relação linear entre as variáveis. No entanto, é crucial entender que isso **não significa necessariamente que não haja *qualquer* tipo de relação** entre elas. Pode existir uma relação não linear (curvilínea, quadrática, etc.) que o coeficiente de Pearson, por ser uma medida de linearidade, não é capaz de capturar. Por exemplo, duas variáveis podem ter uma relação parabólica; neste caso, a correlação de Pearson pode ser próxima de zero, mas a relação ainda existe e é clara.

## Fórmulas

### Fórmula da Regressão Linear Simples

A equação da reta de regressão linear simples é:

$$ \hat{Y} = b_0 + b_1 X $$

Onde:
*   $ \hat{Y} $ é o valor previsto da variável dependente.
*   $ b_0 $ é o intercepto (valor de Y quando X é 0).
*   $ b_1 $ é o coeficiente angular (mudança em Y para cada unidade de mudança em X).
*   $ X $ é a variável independente.

Os coeficientes $ b_0 $ e $ b_1 $ são calculados usando o método dos mínimos quadrados:

$$ b_1 = \frac{n\sum(XY) - \sum X \sum Y}{n\sum(X^2) - (\sum X)^2} $$

$$ b_0 = \bar{Y} - b_1 \bar{X} $$

Onde:
*   $ n $ é o número de observações.
*   $ \sum XY $ é a soma do produto de X e Y.
*   $ \sum X $ é a soma de X.
*   $ \sum Y $ é a soma de Y.
*   $ \sum(X^2) $ é a soma dos quadrados de X.
*   $ \bar{X} $ é a média de X.
*   $ \bar{Y} $ é a média de Y.

### Fórmula do Coeficiente de Correlação de Pearson (r)

O coeficiente de correlação de Pearson é calculado por:

$$ r = \frac{n\sum(XY) - \sum X \sum Y}{\sqrt{[n\sum(X^2) - (\sum X)^2][n\sum(Y^2) - (\sum Y)^2]}} $$

Onde:
*   $ n $ é o número de pares de dados.
*   $ \sum XY $ é a soma do produto de X e Y.
*   $ \sum X $ é a soma de X.
*   $ \sum Y $ é a soma de Y.
*   $ \sum(X^2) $ é a soma dos quadrados de X.
*   $ \sum(Y^2) $ é a soma dos quadrados de Y.

```python
from scipy.stats import linregress
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Ensure df is loaded and df_regress is created for regression

# Create a temporary DataFrame without NaNs in the columns of interest for regression
df_regress = df.dropna(subset=['lipideos_g', 'proteina_g'])

# Define the independent (X) and dependent (Y) variables from the cleaned DataFrame
X_regress = df_regress['lipideos_g']
Y_regress = df_regress['proteina_g']

# Perform linear regression
lin_regress_result = linregress(X_regress, Y_regress)
slope = lin_regress_result.slope
intercept = lin_regress_result.intercept
r_value = lin_regress_result.rvalue
p_value = lin_regress_result.pvalue
stderr = lin_regress_result.stderr

print(f"Resultados da Regressão Linear entre Lipídeos (g) e Proteína (g):")
print(f"  Inclinação (Slope): {slope:.4f}")
print(f"  Intercepto: {intercept:.4f}")
print(f"  Coeficiente de Correlação (r): {r_value:.4f}")
print(f"  Valor-p: {p_value:.4f}")
print(f"  Erro Padrão da Inclinação: {stderr:.4f}")

# Adding the regression line to the plot for visual confirmation
plt.figure(figsize=(10, 6))
sns.scatterplot(x='lipideos_g', y='proteina_g', data=df)
# Plot the regression line using the min and max of the *cleaned* independent variable for its range
min_x_regress = X_regress.min()
max_x_regress = X_regress.max()
plt.plot(
    [min_x_regress, max_x_regress],
    [slope * min_x_regress + intercept, slope * max_x_regress + intercept],
    color='blue', linestyle='-', label=f'Linha de Regressão Linear (y = {slope:.2f}x + {intercept:.2f})'
)

# Original 45-degree line for reference (using values from df for full range visualization)
min_val_plot = min(df['lipideos_g'].min(), df['proteina_g'].min())
max_val_plot = max(df['lipideos_g'].max(), df['proteina_g'].max())
plt.plot([min_val_plot, max_val_plot], [min_val_plot, max_val_plot], color='red', linestyle='--', label='Linha de 45 graus (y=x)')

plt.title('Relação entre Lipídeos (g) e Proteína (g) com Linha de Regressão')
plt.xlabel('Lipídeos (g)')
plt.ylabel('Proteína (g)')
plt.grid(True)
plt.legend()
plt.show()
```

#Inclinação (Slope)
A inclinação representa quanto a variável dependente, neste caso, proteína (g/100 g) tende a variar quando a variável independente lipídeos (g/100 g) aumenta em uma unidade. O valor encontrado foi 0,2287. Portanto, para cada aumento de 1 g/100 g de lipídeos, espera-se, em média, um aumento de aproximadamente 0,23 g/100 g de proteína.

#Intercepto
O intercepto é o valor que a reta de regressão prevê para a variável Y quando X é igual a zero. Nesse caso, quando o teor de lipídeos é 1 g/100 g, o modelo estima um teor de aproximadamente 8,31 g/100 g de proteína. A equação da reta pode ser representada por:

Proteína = 8,3065 + 0,2287 × Lipídeos

#Coeficiente de correlação de Pearson (r)
O coeficiente de Pearson mede a força e a direção da relação linear entre duas variáveis. O resultado foi r = 0,2799, ou aproximadamente 0,28. Como o valor é positivo, existe uma tendência de que maiores valores de lipídeos estejam associados a maiores valores de proteína.

#R²
O R² (R-quadrado) mostra quanto da variação observada na variável dependente pode ser explicada pelo modelo de regressão utilizando a variável independente. Como:

R² = r² = (0,2799)² ≈ 0,078

temos aproximadamente 7,8%.

Isso significa que, dentro desse modelo, cerca de 7,8% da variação observada no teor de proteína pode ser explicada pela variação no teor de lipídeos. Consequentemente, aproximadamente 92,2% da variação restante não é explicada pelos lipídeos nesse modelo.

#Valor-p
O valor-p é utilizado para avaliar a evidência estatística de uma relação linear entre as duas variáveis. O resultado apresentado foi 0,0000, indicando um valor-p extremamente pequeno. Considerando o nível de significância usual de 5% (0,05), o resultado é estatisticamente significativo.

Isso significa que há evidências estatísticas de uma relação linear entre o teor de lipídeos e o teor de proteína nos dados analisados.

#Erro padrão da inclinação
O erro padrão da inclinação indica a incerteza associada à estimativa do coeficiente angular da reta. No caso, a inclinação estimada foi 0,2287, com erro padrão de 0,0326. De maneira geral, quanto menor o erro padrão em relação à estimativa da inclinação, maior tende a ser a precisão dessa estimativa. Assim, o resultado indica que a estimativa de 0,2287 possui uma determinada precisão estatística e que existe evidência de uma associação positiva entre as variáveis.

#Proteina seria uma boa variavel para explicar os lipídeos?
Em resumo, enquanto existe uma relação estatisticamente detectável e positiva entre lipídeos e proteínas, essa relação é muito fraca e a variável lipídeos explica uma fração muito pequena da variância da proteína. Portanto, a proteína por si só não seria um bom preditor ou explicador significativo do teor de lipídeos neste conjunto de dados.
