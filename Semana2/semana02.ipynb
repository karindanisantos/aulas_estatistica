```python
import pandas as pd
df = pd.read_excel('dados_semana02.xlsx')
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Contar a frequência de cada categoria da variável 'Sexo'
sexo_counts = df['Sexo'].value_counts()

# Criar o gráfico de pizza
plt.figure(figsize=(4, 4))
plt.pie(sexo_counts, labels=sexo_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel', len(sexo_counts)))
plt.suptitle ('Gráfico de Pizza - Variavel Categórica Nominal', fontsize=16, y= 1.02, fontweight="bold")
plt.title('Distribuição de Sexo dos Alunos')
plt.axis('equal') # Garante que o gráfico de pizza seja um círculo.
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Peso_kg'])
plt.suptitle('Gráfico Boxplot: Peso (kg) - Variável Quantitativa Contínua', fontsize=18, y=1.10, fontweight='bold')
plt.title('Distribuição do Peso dos Alunos', fontsize=16, pad=15)
plt.xlabel('Peso (kg)', fontsize=14, labelpad=10)
plt.ylabel('Valores', fontsize=12)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.countplot(data=df, x='Numero_disciplinas', hue='Numero_disciplinas', palette='viridis', legend=False)
plt.suptitle('Gráfico de Barras: Número de Disciplinas - Variável Quantitativa discreta', fontsize= 16, y=1.06, fontweight='bold')
plt.title('Distribuição do Número de Disciplinas por Aluno', fontsize=13, pad=15)
plt.xlabel('Número de Disciplinas', fontsize=14, labelpad=10)
plt.ylabel('Contagem', fontsize=12)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Criar o histograma para a variável 'Altura_m'
#Histograma do Tipo Variavel quantitativa continua
plt.figure(figsize=(8, 4))
sns.histplot(df['Altura_m'], bins=10, kde=True)
plt.suptitle (' Histograma - Variavel Quantitativa Continua', fontsize=20,  y= 1.10, fontweight="bold")
plt.title('Distribuição da Altura (m)', fontsize= 18, pad= 16 )
plt.xlabel('Altura (m)',fontsize= 16)
plt.ylabel('Frequência',fontsize= 16)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

#Grafico Extra
O histograma e o boxplot foram construídos utilizando a variável altura, permitindo comparar diferentes formas de representação da distribuição dos mesmos dados. O histograma permite visualizar a frequência dos valores em diferentes intervalos, enquanto o boxplot resume a distribuição por meio da mediana, quartis e possíveis valores discrepantes.

```python
plt.figure(figsize=(8, 4))
sns.boxplot(x=df['Altura_m'])
plt.suptitle('Gráfico Boxplot: Altura (m) - Variável Quantitativa Contínua', fontsize=18, y=1.10, fontweight='bold')
plt.title('Distribuição da Altura dos Alunos', fontsize=16, pad=15)
plt.xlabel('Altura (m)', fontsize=14, labelpad=10)

plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```
