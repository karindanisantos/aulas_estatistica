```python
import pandas as pd
df = pd.read_excel('dados_semana2.xlsx')

```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Criar o histograma para a variável 'Altura_m'
#Histograma do Tipo Variavel quantitativa continua
plt.figure(figsize=(8, 4))
sns.histplot(df['Altura_m'], bins=10, kde=True)
plt.suptitle (' Histograma do Tipo Variavel Quantitativa Continua', fontsize=20,  y= 1.10, fontweight="bold")
plt.title('Distribuição da Altura (m)', fontsize= 18, pad= 16 )
plt.xlabel('Altura (m)',fontsize= 16)
plt.ylabel('Frequência',fontsize= 16)
plt.grid(axis='y', alpha=0.75)
```

```python
plt.figure(figsize=(8,4))
sns.countplot(data=df, x='Numero_disciplinas', hue='Numero_disciplinas', palette='viridis', legend=False)
plt.suptitle('Gráfico de Barras: Número de Disciplinas', fontsize=16, y=1.06, fontweight='bold')
plt.title('Distribuição do Número de Disciplinas por Aluno', fontsize=13, pad=15)
plt.xlabel('Número de Disciplinas', fontsize=14, labelpad=10)
plt.ylabel('Contagem', fontsize=12)
plt.xticks(fontsize=12)
plt.yticks(fontsize=12)
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
# Contar a frequência de cada nível de satisfação
satisfacao_counts = df['Satisfacao'].value_counts()

# Criar o gráfico de pizza
plt.figure(figsize=(5, 5))
plt.pie(satisfacao_counts, labels=satisfacao_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('viridis', len(satisfacao_counts)))
plt.suptitle ('Gráfico de Pizza Variavel Ordinal', fontsize=16,  y= 1.00, fontweight="bold")
plt.title('Distribuição da Satisfação dos Alunos')
plt.axis('equal') # Garante que o gráfico de pizza seja um círculo.
plt.show()
```

```python
# Contar a frequência de cada categoria da variável 'Sexo'
sexo_counts = df['Sexo'].value_counts()

# Criar o gráfico de pizza
plt.figure(figsize=(5, 5))
plt.pie(sexo_counts, labels=sexo_counts.index, autopct='%1.1f%%', startangle=90, colors=sns.color_palette('pastel', len(sexo_counts)))
plt.suptitle ('Gráfico de Pizza Variavel Nominal', fontsize=16,  y= 1.00, fontweight="bold")
plt.title('Distribuição de Sexo dos Alunos')
plt.axis('equal') # Garante que o gráfico de pizza seja um círculo.
plt.show()
```

Com base na estrutura do seu DataFrame e nos tipos de dados, as seguintes colunas podem ser consideradas **variáveis discretas** (ou categorizadas como tal para fins de análise, mesmo que o tipo de dado subjacente possa ser numérico):

*   **`ID`**: É um identificador único, discreto.
*   **`Sexo`**: Variável categórica nominal (F, M).
*   **`Curso`**: Variável categórica nominal.
*   **`Periodo`**: Variável categórica nominal (Manhã, Tarde, Noite).
*   **`Numero_disciplinas`**: É uma contagem de disciplinas, sendo uma variável quantitativa discreta.
*   **`Horas_estudo_semana`**: É uma contagem de horas, também uma variável quantitativa discreta.
*   **`Satisfacao`**: Variável categórica ordinal (Baixa, Média, Alta).
