## Análise dos Modelos Probabilísticos

Vamos abordar cada uma das situações propostas, definindo a variável aleatória, identificando os possíveis valores, escolhendo o modelo probabilístico mais adequado, determinando seus parâmetros e justificando a escolha.

### Situação A: Inspeção de uma embalagem de alimento

*   **1. Variável Aleatória de Interesse:**
    *   $X$: Resultado da inspeção de uma única embalagem.

*   **2. Possíveis Valores:**
    *   $X = 1$ (se a embalagem é adequada)
    *   $X = 0$ (se a embalagem é inadequada)

*   **3. Modelo Mais Adequado:**
    *   **Bernoulli**

*   **4. Parâmetro(s) do Modelo:**
    *   $p$: Probabilidade de sucesso (embalagem adequada). Neste caso, poderíamos definir $p$ como a probabilidade de ser adequada ou a probabilidade de ser inadequada, dependendo do foco. Se definirmos 'sucesso' como 'embalagem adequada', então $p$ é essa probabilidade. Se definirmos 'sucesso' como 'embalagem inadequada', então $p$ seria a probabilidade de ser inadequada (não fornecida, mas geralmente assumida como o evento de interesse para falhas).

*   **5. Justificativa da Escolha:**
    *   A distribuição de Bernoulli é apropriada para experimentos com apenas dois resultados possíveis (sucesso ou falha), onde cada tentativa é independente e tem uma probabilidade constante de sucesso. Neste caso, a inspeção de uma única embalagem é uma única tentativa que pode resultar em 'adequada' ou 'inadequada'.
*  **6. Expressão Probabilística:**
    * $P(X=k) = p^k (1-p)^{1-k}$$P(X=k) = p^k (1-p)^{1-k}$, onde $k \in \{0, 1\}$$k \in \{0, 1\}$.


### Situação B: Número de embalagens inadequadas em 30 selecionadas

*   **1. Variável Aleatória de Interesse:**
    *   $Y$: Número de embalagens inadequadas em 30 embalagens selecionadas.

*   **2. Possíveis Valores:**
    *   $Y = 0, 1, 2, ..., 30$ (inteiros de 0 a 30)

*   **3. Modelo Mais Adequado:**
    *   **Binomial**

*   **4. Parâmetro(s) do Modelo:**
    *   $n = 30$: Número de tentativas (embalagens selecionadas).
    *   $p = 0.05$: Probabilidade de sucesso (probabilidade de uma única embalagem ser inadequada).

*   **5. Justificativa da Escolha:**
    *   A distribuição Binomial é usada quando há um número fixo de tentativas independentes ($n$), cada uma com apenas dois resultados possíveis (sucesso/falha), e a probabilidade de sucesso ($p$) é constante para cada tentativa. Aqui, 'sucesso' é uma embalagem ser inadequada, e estamos contando o número de 'sucessos' em 30 tentativas.
*   **6. Expressão Probabilística:**
    *  $P(Y=k) = \binom{n}{k} p^k (1-p)^{n-k}$$P(Y=k) = \binom{n}{k} p^k (1-p)^{n-k}$, onde $k \in \{0, 1, ..., 30\}$$k \in \{0, 1, ..., 30\}$.

### Situação C: Número de falhas de equipamento durante uma hora

*   **1. Variável Aleatória de Interesse:**
    *   $Z$: Número de falhas de determinado equipamento durante uma hora de produção.

*   **2. Possíveis Valores:**
    *   $Z = 0, 1, 2, 3, ...$ (inteiros não negativos)

*   **3. Modelo Mais Adequado:**
    *   **Poisson**

*   **4. Parâmetro(s) do Modelo:**
    *   $\lambda = 2$: Taxa média de ocorrências por unidade de tempo (2 falhas por hora).

*   **5. Justificativa da Escolha:**
    *   A distribuição de Poisson é apropriada para modelar o número de eventos que ocorrem em um intervalo fixo de tempo ou espaço, dado que esses eventos ocorrem com uma taxa média constante ($\lambda$) e independentemente do tempo desde o último evento. Aqui, estamos contando o número de falhas em um período de uma hora com uma taxa média conhecida.

*   **6. Expressão Probabilística:**
    * $P(Z=k) = \frac{e^{-\lambda} \lambda^k}{k!}$$P(Z=k) = \frac{e^{-\lambda} \lambda^k}{k!}$, onde $k \in \{0, 1, 2, ...\}$$k \in \{0, 1, 2, ...\}$.

### 6. Comparação das três situações

A principal diferença entre as três situações reside no tipo de evento que está sendo observado e contado:

*   **Situação A (Bernoulli):** Observamos o resultado de **uma única tentativa** com dois resultados possíveis (dicotômica). É a unidade fundamental de um experimento binário.

*   **Situação B (Binomial):** Observamos o **número de sucessos em um número fixo de tentativas independentes** (contagem de eventos discretos em um `n` fixo). É uma extensão da Bernoulli para múltiplas tentativas.

*   **Situação C (Poisson):** Observamos o **número de ocorrências de um evento em um intervalo contínuo de tempo ou espaço** (contagem de eventos em um intervalo). O número de tentativas não é fixo e é potencialmente infinito, e a probabilidade de ocorrência em um ponto específico é muito pequena, mas a taxa média ao longo do intervalo é conhecida.

### 7. Para a Situação B, calcule a probabilidade de exatamente duas das 30 embalagens serem inadequadas.

*   **Variável Aleatória:** $Y \sim Binomial(n=30, p=0.05)$
*   **Cálculo:** $P(Y = 2) = \binom{30}{2} (0.05)^2 (1 - 0.05)^{30 - 2}$
*   Usaremos a função `binom.pmf` do módulo `scipy.stats`.

```python
from scipy.stats import binom

n = 30  # Número de embalagens
p = 0.05 # Probabilidade de uma embalagem ser inadequada
k = 2   # Número exato de embalagens inadequadas

prob_b = binom.pmf(k, n, p)
print(f"A probabilidade de exatamente duas das 30 embalagens serem inadequadas é: {prob_b:.4f}")

# Interpretação do resultado
print(f"Isso significa que, se inspecionarmos 30 embalagens muitas vezes, esperaríamos que em aproximadamente {prob_b*100:.2f}% das vezes, exatamente duas delas estivessem inadequadas.")
```

### 8. Para a Situação C, calcule a probabilidade de nenhuma falha ser observada durante uma hora.

*   **Variável Aleatória:** $Z \sim Poisson(\lambda=2)$
*   **Cálculo:** $P(Z = 0) = \frac{e^{-\lambda} \lambda^0}{0!} = e^{-2}$
*   Usaremos a função `poisson.pmf` do módulo `scipy.stats`.

```python
from scipy.stats import poisson

lambda_c = 2      # Taxa média de falhas por hora
k_c = 0           # Número de falhas (nenhuma falha)

prob_c = poisson.pmf(k_c, lambda_c)
print(f"A probabilidade de nenhuma falha ser observada durante uma hora é: {prob_c:.4f}")

# Interpretação do resultado
print(f"Isso significa que, em um longo período de tempo, esperaríamos que em aproximadamente {prob_c*100:.2f}% das horas de produção, não ocorresse nenhuma falha no equipamento.")
```
