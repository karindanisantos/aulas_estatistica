### Calculando a Correlação Estatística

````python
correlation = plot_df['Temperatura de armazenamento'].corr(plot_df['Perda de massa (%)'])
print(f"A correlação estatística entre Temperatura de armazenamento e Perda de massa (%) é: {correlation:.2f}")
````

A correlação encontrada foi de 1.00. Isso significa que a correlação é positiva e perfeita.

Positiva: Indica que, à medida que a 'Temperatura de armazenamento' aumenta, a 'Perda de massa (%)' também aumenta. As duas variáveis se movem na mesma direção.
Perfeita (1.00): Significa que existe uma relação linear exata entre as duas variáveis. Para cada aumento na temperatura, há um aumento proporcional e consistente na perda de massa, sem desvios da linha de tendência. Ou seja, todos os pontos de dados (os dois que temos) estão exatamente na linha de regressão.

Ausência de Relação Linear: O mais importante é que não há uma tendência clara e linear entre as duas variáveis. Ou seja, não podemos dizer que, à medida que uma variável aumenta, a outra aumenta ou diminui consistentemente em uma linha reta.

Não Implica Ausência Total de Relação: É crucial entender que uma correlação próxima de zero não significa que não há qualquer relação entre as variáveis. Pode haver uma relação, mas ela pode ser:

Não linear: As variáveis podem ter uma relação curvilínea (por exemplo, uma parábola), onde a correlação de Pearson (que mede apenas a relação linear) não seria capaz de capturá-la. Por exemplo, uma variável pode aumentar, depois diminuir, e depois aumentar novamente conforme a outra variável muda.
Qualitativa: A relação pode ser complexa e depender de outras variáveis ou ser de natureza qualitativa, não capturável por uma métrica linear simples.
Em resumo, uma correlação de Pearson próxima de zero nos diz que uma linha reta não é um bom modelo para descrever a relação entre as variáveis, mas um padrão mais complexo ainda pode existir

### Calculando a Regressão Linear

````python
from scipy import stats

# Perform linear regression
slope, intercept, r_value, p_value, std_err = stats.linregress(
    plot_df['Temperatura de armazenamento'],
    plot_df['Perda de massa (%)']
)

print(f"Parâmetros da Regressão Linear:")
print(f"  Inclinação (Slope): {slope:.2f}")
print(f"  Intercepto (Intercept): {intercept:.2f}")
print(f"  Coeficiente de Correlação (r-value): {r_value:.2f}")
print(f"  R-quadrado (R-squared): {r_value**2:.2f}")
print(f"  Valor-p (P-value): {p_value:.3f}")
print(f"  Erro Padrão (Standard Error): {std_err:.2f}")
````

### Fórmula da Regressão Linear Simples

A regressão linear simples modela a relação entre duas variáveis (uma variável dependente $Y$ e uma variável independente $X$) como uma linha reta. A fórmula é:

$$ \hat{Y} = b_0 + b_1 X $$

Onde:
*   $ \hat{Y} $ é o valor predito da variável dependente.
*   $ b_0 $ é o intercepto (o valor de $ \hat{Y} $ quando $ X = 0 $).
*   $ b_1 $ é a inclinação (slope), que representa a mudança média em $ \hat{Y} $ para cada unidade de mudança em $ X $.
*   $ X $ é o valor da variável independente.

Os coeficientes $ b_0 $ e $ b_1 $ são calculados usando o método dos mínimos quadrados:

$$ b_1 = \frac{n\sum(XY) - \sum X \sum Y}{n\sum(X^2) - (\sum X)^2} $$

$$ b_0 = \bar{Y} - b_1 \bar{X} $$

### Fórmula do Coeficiente de Correlação de Pearson

O coeficiente de correlação de Pearson ($r$) mede a força e a direção de uma relação linear entre duas variáveis quantitativas. Seu valor varia de -1 a +1.

$$ r = \frac{n\sum(XY) - \sum X \sum Y}{\sqrt{\left(n\sum(X^2) - (\sum X)^2\right)\left(n\sum(Y^2) - (\sum Y)^2\right)}} $$

Onde:
*   $ n $ é o número de pares de dados.
*   $ \sum XY $ é a soma dos produtos dos valores de $X$ e $Y$.
*   $ \sum X $ e $ \sum Y $ são as somas dos valores de $X$ e $Y$, respectivamente.
*   $ \sum X^2 $ e $ \sum Y^2 $ são as somas dos quadrados dos valores de $X$ e $Y$, respectivamente.

### Explicação da Regressão Linear Simples

A Regressão Linear Simples é uma técnica estatística utilizada para modelar a relação entre duas variáveis quantitativas: uma variável dependente (ou resposta), que queremos prever, e uma variável independente (ou preditora), que usamos para fazer a previsão. O objetivo é encontrar a linha reta que melhor se ajusta aos dados, minimizando a distância entre os pontos de dados e a linha.

Neste estudo, a **Perda de massa (%)** é a variável dependente (Y), e a **Temperatura de armazenamento (°C)** é a variável independente (X). A equação da linha de regressão é dada por:

$$ \hat{Y} = b_0 + b_1 X $$

Onde:
*   $ \hat{Y} $: É o valor predito da perda de massa (%).
*   $ X $: É o valor da temperatura de armazenamento (°C).
*   $ b_0 $ (Intercepto): Representa o valor médio da perda de massa (%) quando a temperatura de armazenamento é zero. Em alguns contextos, como este, um intercepto zero pode não ter um significado físico direto, pois a perda de massa a 0°C pode ser diferente, ou 0°C pode estar fora do escopo prático dos dados. Em nosso caso, o intercepto de **-12.43** indica que, se a tendência linear se estendesse a 0°C, a perda de massa seria negativa, o que fisicamente não é possível, sugerindo que o modelo é válido dentro da faixa de temperaturas observadas.
*   $ b_1 $ (Inclinação ou Slope): Representa o quanto a perda de massa (%) muda, em média, para cada aumento de uma unidade na temperatura de armazenamento (°C). Uma inclinação positiva significa que a perda de massa aumenta com a temperatura, enquanto uma inclinação negativa significaria o contrário. Nosso valor de inclinação de **1.25** significa que, para cada grau Celsius que a temperatura de armazenamento aumenta, a perda de massa das mangas aumenta em aproximadamente 1.25 pontos percentuais.

#### Coeficiente de Correlação (r-value) e R-quadrado (R-squared)

*   **Coeficiente de Correlação (r-value)**: Mede a força e a direção da relação linear entre duas variáveis. Varia de -1 a +1. Um valor de **+1.00** como o que encontramos indica uma correlação linear positiva perfeita. Isso significa que, sem exceção, à medida que a temperatura de armazenamento aumenta, a perda de massa (%) também aumenta, e todos os pontos de dados (as duas observações) se alinham perfeitamente em uma linha reta.

*   **R-quadrado (R-squared)**: Também conhecido como coeficiente de determinação, indica a proporção da variância na variável dependente que pode ser prevista a partir da variável independente. Varia de 0 a 1. Um R-quadrado de **1.00** significa que 100% da variação na perda de massa (%) é explicada pela variação na temperatura de armazenamento. Este valor perfeito é um reflexo direto de termos apenas dois pontos de dados, que sempre definirão uma linha reta perfeita. Em conjuntos de dados maiores e mais complexos, um R-quadrado de 1.00 é extremamente raro e pode indicar overfitting ou um modelo excessivamente simplificado.
