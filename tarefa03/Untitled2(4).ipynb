```python
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

```python
import plotly.express as px # ver os dados de forma interativa
import pandas as pd



dados = pd.DataFrame({
    "id": range(1, 21),
    "sexo": ["F", "M"] * 10,  # ordinal
    "Preferencia_Chocolate": [

"Branco", "Ao leite", "Meio Amargo", "Amargo",
    "Ao leite", "Branco", "Meio Amargo", "Ao leite",
    "Amargo", "Branco", "Ao leite", "Meio Amargo",
    "Branco", "Amargo", "Ao leite", "Meio Amargo",
    "Ao leite", "Branco", "Amargo", "Meio Amargo"

    ],  # nominal
    "tipo_bebida": [
    "Café", "Chá", "Suco", "Refrigerante", "Água",
    "Café", "Água", "Suco", "Chá", "Café",
    "Refrigerante", "Água", "Café", "Suco", "Chá",
    "Café", "Água", "Refrigerante", "Suco", "Café"
    ], #ordinal

    "numero_frutas_dia": [1, 2, 3, 0, 2, 1, 4, 1, 3, 2, 0, 2, 1, 3, 2, 4, 1, 2, 3, 1],  # discreta
#  quantitativa continua

    "Altura_m": [
        1.62, 1.70, 1.58, 1.75, 1.68,
        1.65, 1.72, 1.60, 1.67, 1.80,
        1.63, 1.69, 1.74, 1.57, 1.71,
        1.66, 1.76, 1.64, 1.73, 1.61
    ],
# quantitativa continua
    "Peso_kg": [
        58, 70, 52, 78, 65,
        60, 72, 55, 63, 82,
        59, 68, 75, 50, 69,
        61, 80, 64, 73, 56
    ]
})

dados
```

### Gráfico de Barras para a variável 'tipo_bebida'

Criação de um gráfico de barras para visualizar a frequência de cada tipo de bebida na base de dados, utilizando a variável ordinal `tipo_bebida`.

O grafico de barras apresenta a distribuição  das preferencias de bebidas entre estudantes.
Observa-se que o café é a bebida mais escolhida pelos estudantes, enquanto o chá e o refrigerante aparecem com menor frequência.

```python
# Calcular a frequência de cada tipo de bebida
freq_tipo_bebida = dados['tipo_bebida'].value_counts().reset_index()
freq_tipo_bebida.columns = ['tipo_bebida', 'count']

# Criar o gráfico de barras
fig = px.bar(
    freq_tipo_bebida,
    x='tipo_bebida',
    y='count',
    title='Frequência de Tipos de Bebida (Variável Ordinal)',
    labels={'tipo_bebida': 'Tipo de Bebida', 'count': 'Contagem'},
    color='tipo_bebida' # Atribuir cores diferentes para cada categoria
)

fig.update_layout(xaxis={'categoryorder':'total descending'}) # Ordenar barras pela contagem
fig.show()
```

### Gráfico de Dispersão: Altura vs. Peso

Criação de um gráfico de dispersão para observar a relação entre a altura e o peso dos indivíduos.

O gráfico de dispersão coloca cada estudante como um ponto, relacionando sua altura com seu peso.

Observa-se uma tendência de aumento do peso conforme a altura aumenta.

```python
fig = px.scatter(
    dados,
    x='Altura_m',
    y='Peso_kg',
    title='Gráfico de Dispersão: Altura (m) vs. Peso (kg)',
    labels={'Altura_m': 'Altura (m)', 'Peso_kg': 'Peso (kg)'},
    hover_data=['id', 'sexo'] # Adicionar informações ao passar o mouse
)

fig.show()
```

Criação de um histograma

O histograma mostra como as alturas estão distribuídas nos diferentes intervalos.

Interpretação: Observa-se que as alturas estão distribuídas principalmente entre 1,60 m e 1,75 m, com poucas observações nos extremos.

```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))

sns.histplot(
    data=dados,
    x="Altura_m",
    bins= [1.57, 1.60, 1.65, 1.70, 1.75, 1.80],
    kde=False,
    color="blue"
)

plt.title("Distribuição da Altura")
plt.xlabel("Altura")
plt.ylabel("Frequência")

plt.show()
```
