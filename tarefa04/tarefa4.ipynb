#Referencial Teórico

**Definição de Quartis**

De acordo com Andrade e Ogliari, os quartis são medidas utilizadas para dividir um conjunto de dados ordenados em quatro partes. O primeiro quartil (Q1) corresponde aproximadamente ao ponto abaixo do qual estão 25% das observações; o segundo quartil (Q2) corresponde à mediana, deixando aproximadamente 50% das observações abaixo dele; e o terceiro quartil (Q3) deixa aproximadamente 75% das observações abaixo dele.

**Definição de Mediana (Q2)**
A mediana é uma medida de tendência central que divide os dados ordenados em duas partes. Metade das observações apresenta valores menores ou iguais à mediana e a outra metade apresenta valores maiores ou iguais. A mediana é utilizada juntamente com quartis na interpretação da distribuição dos dados.

#Calculo do Quartis da Variavl Proteina

```python
print('--- Quartis da variável "Proteina_g" ---')
Q1 = df['Proteina_g'].quantile(0.25)
Q2 = df['Proteina_g'].quantile(0.50)  # Mediana
Q3 = df['Proteina_g'].quantile(0.75)

print(f"Primeiro Quartil (Q1): {Q1:.2f} g")
print(f"Segundo Quartil (Q2 - Mediana): {Q2:.2f} g")
print(f"Terceiro Quartil (Q3): {Q3:.2f} g")
```

 #Boxplot do teor de proteína dos alimento com os valores do calculo de Quartis

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.boxplot(x=df['Proteina_g'])

# Adicionar linhas para os quartis
plt.axvline(Q1, color='r', linestyle='--', label=f'Q1: {Q1:.2f} g')
plt.axvline(Q2, color='g', linestyle='--', label=f'Q2 (Mediana): {Q2:.2f} g')
plt.axvline(Q3, color='b', linestyle='--', label=f'Q3: {Q3:.2f} g')

# Adicionar texto para os quartis (opcional, para maior clareza)
plt.text(Q1, plt.ylim()[1] * 0.95, f'Q1: {Q1:.2f}', color='r', ha='center', va='bottom', bbox=dict(facecolor='white', alpha=0.7))
plt.text(Q2, plt.ylim()[1] * 0.90, f'Q2: {Q2:.2f}', color='g', ha='center', va='bottom', bbox=dict(facecolor='white', alpha=0.7))
plt.text(Q3, plt.ylim()[1] * 0.85, f'Q3: {Q3:.2f}', color='b', ha='center', va='bottom', bbox=dict(facecolor='white', alpha=0.7))

plt.title("Boxplot do teor de proteína dos alimentos com Quartis")
plt.xlabel("Proteína (g)")
plt.legend()
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

#Definição de Boxplot e interpretação do Boxplot de Proteina_g

#O que é um Boxplot?

O boxplot, também denominado diagrama de caixa, é uma **representação gráfica utilizada para descrever a distribuição de uma variável quantitativa**. De acordo com Andrade e Ogliari (2013), a análise exploratória dos dados utiliza medidas como a mediana e os quartis para caracterizar a distribuição dos valores. O boxplot reúne essas informações graficamente, permitindo observar a posição central, a dispersão dos dados e a presença de possíveis observações atípicas.

O gráfico é construído a partir dos quartis e da mediana:

Primeiro quartil (Q1): corresponde ao valor que separa aproximadamente os 25% menores valores dos demais.
Segundo quartil (Q2 ou mediana): divide os dados ordenados em duas partes, deixando aproximadamente 50% das observações abaixo e 50% acima desse valor.
Terceiro quartil (Q3): corresponde ao valor que separa aproximadamente os 75% menores valores dos 25% maiores.
Amplitude interquartílica (AIQ): corresponde à diferença entre o terceiro e o primeiro quartil:
AIQ=Q3−Q1

A caixa do boxplot é delimitada por Q1 e Q3, portanto representa os 50% centrais da distribuição. A linha localizada dentro da caixa representa a mediana (Q2). Os bigodes se estendem até os valores que permanecem dentro dos limites estabelecidos para a identificação de observações atípicas. Valores que ultrapassam esses limites podem ser representados individualmente no gráfico.

Para identificar observações atípicas, utiliza-se o critério baseado na amplitude interquartílica:

LI=Q1−1,5×AIQ
LS=Q3+1,5×AIQ

Assim, valores menores que o limite inferior ou maiores que o limite superior são considerados observações atípicas pelo critério de 1,5 × AIQ.

AIQ=Q3−Q1

AIQ=18,5018−1,1630

#Como determinamos se há observações atípicas em uma base de dados?
Para identificar possiveis valores discrepantes, utilizamos os quartis e distancia interquartílica. Primeiro calcula-se a distancia interquartílica, em seguida sao determinados os limites inferiores e superiores.
### Identificando Observações Atípicas (Outliers) usando a Regra do IQR

A regra do Intervalo Interquartílico (IQR) é um método comum para detectar outliers (Observações Atípicas) em um conjunto de dados. Ela se baseia na ideia de que pontos de dados que estão muito distantes dos quartis (Q1 e Q3) são considerados atípicos.

Os limites para identificar outliers são calculados da seguinte forma:

*   **Limite Inferior (LI)**: `Q1 - 1.5 * IQR`
*   **Limite Superior (LS)**: `Q3 + 1.5 * IQR`

Onde:
*   `Q1` é o Primeiro Quartil (25º percentil)
*   `Q3` é o Terceiro Quartil (75º percentil)
*   `IQR` é a Amplitude Interquartílica (`Q3 - Q1`)

Qualquer ponto de dados que esteja abaixo do Limite Inferior ou acima do Limite Superior é considerado um outlier (Observação Atípica ).
Esse procedimento é especialmente útil porque permite identificar valores muito afastados da região central dos dados sem depender da média ou do desvio-padrão.

```python
# Calcular os limites para identificação de outliers usando a regra do IQR
LI = Q1 - 1.5 * AIQ
LS = Q3 + 1.5 * AIQ

print(f"Limite Inferior para Outliers: {LI:.2f} g")
print(f"Limite Superior para Outliers: {LS:.2f} g")

# Identificar os outliers na variável 'Proteina_g'
outliers = df[(df['Proteina_g'] < LI) | (df['Proteina_g'] > LS)]

print(f"\nNúmero de observações atípicas (outliers) encontradas: {len(outliers)}")

if not outliers.empty:
    print("\nObservações atípicas (outliers) no teor de Proteína (g):")
    display(outliers[['Alimento', 'Proteina_g']])
else:
    print("\nNão foram encontradas observações atípicas usando a regra do IQR para a variável 'Proteina_g'.")
```

**Distribuição Homogênea (em relação ao IQR):** Indica que a maioria dos valores de 'Proteina_g' está concentrada de forma relativamente próxima à mediana e dentro dos limites esperados, definidos pelos quartis. Não há valores que se afastem drasticamente da tendência central da distribuição.
